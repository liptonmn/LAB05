### 1.0. Budget Comparison Challenge

You're organizing an event. Your current budget is $1,000.

- You expect 75 attendees.
- Each attendee costs $12.25 to serve (including food and materials).
- Calculate the **total cost** of the event.
- Compare the total cost to your budget.
- Print a message that says whether you're **within budget** or **over budget**.

➡️ Use your result to answer the quiz question: Are you over or under budget?

````python
attendees = 75
cost_per_person = 12.25
budget = 1000

total_cost = attendees * cost_per_person
is_within_budget = total_cost <= budget

if is_within_budget:
    print("You're within budget!")
else:
    print("You're over budget!")

# Calculate total cost
# Compare to budget and print result
````

### 1.2. Pizza Cost per Square Inch Comparison

You're deciding between two pizzas:

- A 13-inch pizza that costs $18
- A 15-inch pizza that costs $20.50

Using the formula for the area of a circle $A = \pi r^2$, compute the **cost per square inch** for both pizzas. Use the equation: **cost per square inch = total cost ÷ area of the pizza.**

➡️ Use your result to answer the quiz: Which pizza is cheaper per square inch?

````python
import math

# Pizza 1: 13-inch diameter, $18
diameter_1 = 13
cost_1 = 18
radius_1 = diameter_1 / 2
area_1 = math.pi * radius_1**2
cost_per_sq_in_1 = cost_1 / area_1

# Pizza 2: 15-inch diameter, $20.50
diameter_2 = 15
cost_2 = 20.50
radius_2 = diameter_2 / 2
area_2 = math.pi * radius_2**2
cost_per_sq_in_2 = cost_2 / area_2

print(f'13-inch pizza: area = {area_1:.2f} sq in, cost per sq in = ${cost_per_sq_in_1:.4f}')
print(f'15-inch pizza: area = {area_2:.2f} sq in, cost per sq in = ${cost_per_sq_in_2:.4f}')

if cost_per_sq_in_1 < cost_per_sq_in_2:
    print('The 13-inch pizza is cheaper per square inch.')
else:
    print('The 15-inch pizza is cheaper per square inch.')

# Use 3.14159 for π
# Calculate and compare the cost per square inch for both pizzas
````
